```python
import pandas as pd

# Basic cleaning
df = df.dropna()             # Remove missing values if any
df = df.drop_duplicates()    # Remove duplicate rows

# Optional: remove outliers if needed
df = df[(df['hours_studied'] <= 12) & (df['sleep_hours'] <= 10)]
```

```python
import matplotlib.pyplot as plt

plt.plot(df['hours_studied'], df['sleep_hours'], color='blue', alpha=0.6)
plt.xlabel('Hours Studied')
plt.ylabel('Sleep Hours')
plt.title('Relationship Between Hours Studied and Sleep Hours')
plt.show()
```

```python
df['study_bins'] = pd.cut(df['hours_studied'], bins=[0,3,6,9,12], labels=['Low','Medium','High','Very High'])
sleep_trends = df.groupby('study_bins')['sleep_hours'].mean()
print(sleep_trends)
sleep_trends.plot(kind='bar', color='skyblue', title='Avg Sleep Hours by Study Range')
plt.show()
```

```python
plt.scatter(df['hours_studied'], df['exam_score'], color='green', alpha=0.6)
plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')
plt.title('Study Hours vs Exam Score')
plt.show()
```

```python
df_sorted = df.sort_values(by='hours_studied')
plt.plot(df_sorted['hours_studied'], df_sorted['sleep_hours'], color='blue', marker='o')
plt.xlabel('Hours Studied')
plt.ylabel('Sleep Hours')
plt.title('Sorted Line Plot: Hours Studied vs Sleep Hours')
plt.show()
```
